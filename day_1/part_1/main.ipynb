```python
puzle_input = open("puzle_input.txt", 'r').read()
rotations = puzle_input.splitlines()
# print(rotations)

# instruction test #1
# rotations = [
#     "R8",
#     "L19"
# ]

# instruction test #2
# rotations = [
#     "L10",
#     "R5"
# ]

# rotations = [
#     "L68",
#     "L30",
#     "R48",
#     "L5",
#     "R60",
#     "L55",
#     "L1",
#     "L99",
#     "R14",
#     "L82"
# ]

def calculate_rotation(current: int, rotation: str) -> int:
    if rotation.startswith('L'):
        # rotate left
        number = int(rotation[1:])
        return perform_calculation(current, number, True)
    
    if rotation.startswith('R'):
        # rotate right
        number = int(rotation[1:])
        return perform_calculation(current, number, False)
    
    print("Rotation: " + rotation)
    raise Exception("unexpected input")

def perform_calculation(current: int, change: int, negative: bool) -> int:
    min=0
    max=99
    
    for i in range(0, change):
        if negative:
            current=current-1
        else:
            current=current+1

        if current < min:
            current=max
        if current > max:
            current=min
    
    return current

current=50
number_of_times_pointing_at_zero=0
for item in rotations:
    current=calculate_rotation(current, item)
    if current == 0:
        number_of_times_pointing_at_zero+=1

print("Current: " + str(current))
print("NOTPAZ: " + str(number_of_times_pointing_at_zero))
```
