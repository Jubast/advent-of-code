```python
from typing import List

puzle_input = open("puzle_input.txt", 'r').read()
id_ranges = puzle_input.strip().split(",")

# instruction test
# id_ranges=[
#     "11-22",
#     "95-115",
#     "998-1012",
#     "1188511880-1188511890",
#     "222220-222224",
#     "1698522-1698528",
#     "446443-446449",
#     "38593856-38593862",
#     "565653-565659",
#     "824824821-824824827",
#     "2121212118-2121212124"
# ]

class Range:
    def __init__(self, start: int, end: int):
        self.start = start
        self.end = end

    def __str__(self):
        return str(self.start) + "-" + str(self.end)


def to_range(id_range: str) -> Range:
    id_splited = id_range.split("-")
    start = int(id_splited[0])
    end = int(id_splited[1])
    return Range(start, end)

def is_number_repeated(number: int) -> bool:
    str_number = str(number)
    length = len(str_number)
    if length % 2 != 0:
        return False
    
    half_length = int(length / 2)
    part_1 = str_number[:half_length]
    part_2 = str_number[half_length:]
    
    if part_1 == part_2:
        return True
    
    return False

def find_repeated_numbers_in_range(id_range: Range) -> List[int]:
    repeated_numbers: List[int] = []
    for number in range(id_range.start, id_range.end + 1):
        if is_number_repeated(number):
            repeated_numbers.append(number)
    return repeated_numbers

def sum_repeated_numbers(repeated_numbers: List[int]) -> int:
    sum = 0
    for repeated_number in repeated_numbers:
        sum += repeated_number
    return sum

all_repeated_numbers: List[int] = []
for id_range in id_ranges:
    parsed_range = to_range(id_range)
    repeated_numbers = find_repeated_numbers_in_range(parsed_range)
    all_repeated_numbers.extend(repeated_numbers)
    
sum = sum_repeated_numbers(all_repeated_numbers)
print(sum)



```
