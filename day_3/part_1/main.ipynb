```python
from typing import List

puzle_input = open("puzle_input.txt", 'r').read()
banks = puzle_input.strip().splitlines()

# banks = [
#     "987654321111111",
#     "811111111111119",
#     "234234234234278",
#     "818181911112111"
# ]

def find_largest_joltage(bank: str) -> int:
    first_largest_value=0
    second_largest_value=0
    length = len(bank)
    for index, battery in enumerate(bank):
        battery_value = int(battery)

        if battery_value > first_largest_value:
            is_last_item = index + 1 == length
            if is_last_item == False:
                first_largest_value = battery_value
                second_largest_value = 0
                continue

        if battery_value > second_largest_value:
            second_largest_value = battery_value
            continue
    
    joltage = str(first_largest_value) + str(second_largest_value)
    int_joltage = int(joltage)
    return int_joltage

def sum_joltage_rating(joltages: List[int]) -> int:
    sum = 0
    for joltage in joltages:
        sum += joltage
    return sum

all_joltages: List[int] = []
for bank in banks:
    joltage = find_largest_joltage(bank)
    all_joltages.append(joltage)

sum = sum_joltage_rating(all_joltages)
print(sum)
```
