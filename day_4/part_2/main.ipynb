```python
from typing import List

puzle_input = open("puzle_input.txt", 'r').read()
lines = puzle_input.strip().splitlines()

# lines = [
#     "..@@.@@@@.",
#     "@@@.@.@.@@",
#     "@@@@@.@.@@",
#     "@.@@@@..@.",
#     "@@.@@@@.@@",
#     ".@@@@@@@.@",
#     ".@.@.@.@@@",
#     "@.@@@.@@@@",
#     ".@@@@@@@@.",
#     "@.@.@@@.@."
# ]

class Map:
    def __init__(self, map: List[List[str]]):
        self.map: List[List[str]] = map

class Index:
    def __init__(self, x: int, y: int):
        self.x: int = x
        self.y: int = y

def number_of_adjacent_rolls_of_paper(map: Map, index: Index) -> int:
    min = 0
    max = len(map.map) - 1

    # if (index.y == 0 and index.x == 7):
    #     print("here")

    count = 0
    for y in range(index.y - 1, index.y + 2):
        if (y < min or y > max):
            continue

        for x in range(index.x -1, index.x + 2):
            if (x < min or x > max):
                continue

            if (index.y == y and index.x == x):
                continue

            if map.map[y][x] == "@":
                count += 1
    
    return count

def can_be_accessed_by_forklift(map: Map, index: Index) -> bool:
    count = number_of_adjacent_rolls_of_paper(map, index)
    if count < 4:
        return True
    return False

def number_of_rolls_of_paper_that_can_be_acceesed(map: Map) -> List[Index]:
    accesed_indexes: List[Index] = []
    for y, line in enumerate(map.map):
        for x, char in enumerate(line):
            if char != "@":
                continue

            index = Index(x, y)
            if can_be_accessed_by_forklift(map, index):
                accesed_indexes.append(index)

    return accesed_indexes

def update_map_with_accessed_indexes(map: Map, accesed_indexes: List[Index]):
    for y, line in enumerate(map.map):
        for x, char in enumerate(line):
            if any(index.x == x and index.y == y for index in accesed_indexes):
                map.map[y][x] = "X"

map_content: List[List[str]] = [[0 for y in range(len(lines))] for x in range(len(lines))]
for y, line in enumerate(lines):
    for x, char in enumerate(line):
        map_content[y][x] = char
map = Map(map_content)

total_count = 0
while True:
    accesed_indexes = number_of_rolls_of_paper_that_can_be_acceesed(map)
    count = len(accesed_indexes)
    if count <= 0:
        break
    total_count += count
    update_map_with_accessed_indexes(map, accesed_indexes)

print(total_count)
```
