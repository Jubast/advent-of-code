```python
from typing import List

puzle_input = open("puzle_input.txt", 'r').read()
lines = puzle_input.strip().splitlines()

# lines = [
#     "3-5",
#     "10-14",
#     "16-20",
#     "12-18",
#     "",
#     "1",
#     "5",
#     "8",
#     "11",
#     "17",
#     "32"
# ]

class Range:
    def __init__(self, min: int, max: int):
        self.min = min
        self.max = max

def parse_input(lines: List[str]) -> tuple[List[Range], List[int]]:
    ranges: List[Range] = []
    ingredients: List[int] = []

    found_blank_line = False
    for line in lines:
        if line == "":
            found_blank_line = True
            continue

        if found_blank_line:
            ingredients.append(int(line))
        else:
            str_range = line.split("-")
            range = Range(int(str_range[0]), int(str_range[1]))
            ranges.append(range)

    # print(ranges)
    # print(ingredients)

    return (ranges, ingredients)

def is_ingredient_in_range(ingredient_range: Range, ingredient_in_question: int) -> bool:
    return ingredient_in_question >= ingredient_range.min  and ingredient_in_question <= ingredient_range.max

def is_ingredient_in_any_range(ingredient_ranges: List[Range], ingredient_in_question: int) -> bool:
    for ingredient_range in ingredient_ranges:
        if is_ingredient_in_range(ingredient_range, ingredient_in_question):
            return True
        
    return False

ranges, ingredients = parse_input(lines)

count = 0
for ingredient in ingredients:
    if is_ingredient_in_any_range(ranges, ingredient):
        count += 1

print(count)


    
```
