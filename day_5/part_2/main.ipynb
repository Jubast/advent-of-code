```python
from typing import List
import uuid

puzle_input = open("puzle_input.txt", 'r').read()
lines = puzle_input.strip().splitlines()

# lines = [
#     "3-5",
#     "10-14",
#     "16-20",
#     "12-18",
#     "",
#     "1",
#     "5",
#     "8",
#     "11",
#     "17",
#     "32"
# ]

class Range:
    def __init__(self, min: int, max: int):
        self.uuid = uuid.uuid4() # oh may xd
        self.min = min
        self.max = max

    def can_be_joined(self, other_range: "Range") -> bool:
        #if self == other_range:
        if self.uuid == other_range.uuid:
            return False
        
        if other_range.min >= self.min and other_range.min <= self.max:
            return True

    def join(self, other_range: "Range") -> "Range":
        new_range = Range(min(self.min, other_range.min), max(self.max, other_range.max))
        return new_range
    
    def __eq__(self, other): 
        if not isinstance(other, Range):
            # don't attempt to compare against unrelated types
            return NotImplemented

        return self.min == other.min and self.max == other.max
    
    def __str__(self):
        return str(self.min) + "-" + str(self.max)


def parse_input(lines: List[str]) -> tuple[List[Range], List[int]]:
    ranges: List[Range] = []
    ingredients: List[int] = []

    found_blank_line = False
    for line in lines:
        if line == "":
            found_blank_line = True
            continue

        if found_blank_line:
            ingredients.append(int(line))
        else:
            str_range = line.split("-")
            range = Range(int(str_range[0]), int(str_range[1]))
            ranges.append(range)

    # print(ranges)
    # print(ingredients)

    return (ranges, ingredients)

def optimize_ranges(ingredient_ranges: List[Range]) -> List[Range]:
    while True:
        found_range_to_combine = False
        for ingredient_range in ingredient_ranges:
            other_range = None
            for r in ingredient_ranges:
                if ingredient_range.can_be_joined(r):
                    other_range = r
            
            if other_range != None:
                found_range_to_combine = True
                combined_range = ingredient_range.join(other_range)
                new_range = []
                for r in ingredient_ranges:
                    if r == ingredient_range or r == other_range:
                        continue
                    new_range.append(r)

                ingredient_ranges = new_range
                ingredient_ranges.append(combined_range)
                break

        if found_range_to_combine == False:
            break

    return ingredient_ranges
        


def number_of_fresh_ingredients_in_range(ingredient_range: Range) -> int:
    count = (ingredient_range.max + 1) - ingredient_range.min
    return count

def number_of_fresh_ingreedients_in_all_ranges(ingredient_ranges: List[Range]) -> int:
    count = 0
    for ingredient_range in ingredient_ranges:
        count += number_of_fresh_ingredients_in_range(ingredient_range)
        
    return count

ranges, ingredients = parse_input(lines)
o_ranges = optimize_ranges(ranges)
count = number_of_fresh_ingreedients_in_all_ranges(o_ranges)

print(count)
# 388342136956631 - ????

    
```
