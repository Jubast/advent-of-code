```python
import math
from typing import List

puzle_input = open("puzle_input.txt", 'r').read()
lines = puzle_input.strip().splitlines()

# lines = [
#     "123 328  51 64 ",
#     " 45 64  387 23 ",
#     "  6 98  215 314",
#     "*   +   *   +  "
# ]

class MathProblem:
    def __init__(self):
        self.numbers: List[int] = []
        self.operand: str = ""

class MathSolution:
    def __init__(self, problem: MathProblem, solution: int):
        self.problem: MathProblem = problem
        self.solution: int = solution

def parse_input(lines: List[str]) -> List[MathProblem]:
    number_of_columns = len(lines[0].strip().split())

    list: List[MathProblem] = [MathProblem() for i in range(number_of_columns)]
    for line in lines:
        striped_line = line.strip()
        splited_line = striped_line.split()
        for i, item in enumerate(splited_line):
            if item.isdigit():
                list[i].numbers.append(int(item))
            else:
                list[i].operand = item

    return list

def solve_math_problems(math_problems: List[MathProblem]) -> List[MathSolution]:
    list: List[MathSolution] = []
    for math_problem in math_problems:
        if math_problem.operand == "*":
            solution = math.prod(math_problem.numbers)
            math_solution = MathSolution(math_problem, solution)
            list.append(math_solution)
        elif math_problem.operand == "+":
            solution = sum(math_problem.numbers)
            math_solution = MathSolution(math_problem, solution)
            list.append(math_solution)
        else:
            raise NotImplementedError()

    return list

math_problems = parse_input(lines)
math_solutions = solve_math_problems(math_problems)

solution_values = [s.solution for s in math_solutions]
print(sum(solution_values))

    
```
