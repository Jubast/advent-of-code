```python
import math
from typing import List

puzle_input = open("puzle_input.txt", 'r').read()
lines = puzle_input.strip().splitlines()

# lines = [
#     "123 328  51 64 ",
#     " 45 64  387 23 ",
#     "  6 98  215 314",
#     "*   +   *   +  "
# ]

class MathProblem:
    def __init__(self):
        self.numbers: List[int] = []
        self.operand: str = ""

class MathSolution:
    def __init__(self, problem: MathProblem, solution: int):
        self.problem: MathProblem = problem
        self.solution: int = solution

def parse_input(lines: List[str]) -> List[MathProblem]:
    math_problems: List[MathProblem] = []

    number_of_chars = len(lines[0])
    rotated_lines: List[str] = ["" for i in range(number_of_chars)]
    for line in lines:
        for i, c in enumerate(line):            
            rotated_lines[i] = rotated_lines[i] + c

    operand = ""
    cephalopod_numbers: List[int] = []
    for rotated_line in rotated_lines:
        if len(rotated_line.strip()) == 0:
            math_problem = MathProblem()
            math_problem.numbers = cephalopod_numbers
            math_problem.operand = operand
            math_problems.append(math_problem)
            # reset
            cephalopod_numbers = []
            operand = ""
            continue

        last_char = rotated_line[-1]
        if last_char.isdigit() or last_char.isspace():
            cephalopod_numbers.append(int(rotated_line))
        else:
            operand = last_char
            cephalopod_numbers.append(int(rotated_line[:-1]))

    # don't forget last item
    math_problem = MathProblem()
    math_problem.numbers = cephalopod_numbers
    math_problem.operand = operand
    math_problems.append(math_problem)

    return math_problems

def solve_math_problems(math_problems: List[MathProblem]) -> List[MathSolution]:
    list: List[MathSolution] = []
    for math_problem in math_problems:
        if math_problem.operand == "*":
            solution = math.prod(math_problem.numbers)
            math_solution = MathSolution(math_problem, solution)
            list.append(math_solution)
        elif math_problem.operand == "+":
            solution = sum(math_problem.numbers)
            math_solution = MathSolution(math_problem, solution)
            list.append(math_solution)
        else:
            raise NotImplementedError()

    return list

math_problems = parse_input(lines)
math_solutions = solve_math_problems(math_problems)

solution_values = [s.solution for s in math_solutions]
print(sum(solution_values))

    
```
