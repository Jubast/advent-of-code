```python
from typing import List

puzle_input = open("puzle_input.txt", 'r').read()
lines = puzle_input.strip().splitlines()

# lines = [
#     ".......S.......",
#     "...............",
#     ".......^.......",
#     "...............",
#     "......^.^......",
#     "...............",
#     ".....^.^.^.....",
#     "...............",
#     "....^.^...^....",
#     "...............",
#     "...^.^...^.^...",
#     "...............",
#     "..^...^.....^..",
#     "...............",
#     ".^.^.^.^.^...^.",
#     "..............."
# ]

first_line = lines[0]
position_of_source = first_line.index("S")

current_beams: List[int] = [0 for i in range(len(first_line))]
current_beams[position_of_source] = 1

# print(first_line)
number_of_splits = 0
for line_index, line in enumerate(lines[1:]):
    next_line_beams: List[int] = [0 for i in range(len(first_line))]

    for beam_index, beam in enumerate(current_beams):
        if beam == 0:
            continue        
        
        if line[beam_index] != "^":
            next_line_beams[beam_index] = 1
            continue

        # splitter found
        number_of_splits += 1
        next_line_beams[beam_index - 1] = 1
        next_line_beams[beam_index + 1] = 1
    
    # simulated_line = list(line)
    # for i, beam in enumerate(next_line_beams):
    #     if beam == 1:
    #         simulated_line[i] = "|"
    # print("".join(simulated_line))

    current_beams = next_line_beams

print(number_of_splits)



    
```
