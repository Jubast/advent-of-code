```python
from typing import List

puzle_input = open("puzle_input.txt", 'r').read()
lines = puzle_input.strip().splitlines()

# lines = [
#     ".......S.......",
#     "...............",
#     ".......^.......",
#     "...............",
#     "......^.^......",
#     "...............",
#     ".....^.^.^.....",
#     "...............",
#     "....^.^...^....",
#     "...............",
#     "...^.^...^.^...",
#     "...............",
#     "..^...^.....^..",
#     "...............",
#     ".^.^.^.^.^...^.",
#     "..............."
# ]

first_line = lines[0]
position_of_source = first_line.index("S")

current_beams: List[int] = [0 for i in range(len(first_line))]
current_beams[position_of_source] = 1

# print(first_line)
# we start with 1 timeline
for line_index, line in enumerate(lines[1:]):
    next_line_beams: List[int] = [0 for i in range(len(first_line))]

    # if line_index == 11:
    #     print("here")

    for beam_index, beam in enumerate(current_beams):
        if beam == 0:
            continue
        
        if line[beam_index] != "^":
            next_line_beams[beam_index] = next_line_beams[beam_index] + beam
            continue

        # splitter found, this default timeline always goes left but
        # but add new timeline if the particle decieds to go right
        next_line_beams[beam_index - 1] = next_line_beams[beam_index - 1] + beam
        next_line_beams[beam_index + 1] = next_line_beams[beam_index + 1] + beam
    
    # simulated_line = list(line)
    # for i, beam in enumerate(next_line_beams):
    #     if beam != 0:
    #         simulated_line[i] = str(beam)
    # print("".join(simulated_line))

    current_beams = next_line_beams

number_of_timelines = [beam for beam in current_beams if beam != 0]

#print(number_of_timelines)
print(sum(number_of_timelines))



    
```
